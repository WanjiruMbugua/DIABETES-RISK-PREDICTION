```python
#Rebin age group

def recode_age(age):
    if 18 <= age <= 29:
        return "18-29"
    elif 30 <= age <= 39:
        return "30-39"
    elif 40 <= age <= 49:
        return "40-49"
    elif 50 <= age <= 59:
        return "50-59"
    elif 60 <= age <= 69:
        return "60-69"
    else:
        return "70+"

Diabetes["Age_group"] = Diabetes["Age"].apply(recode_age)
```

### Strategic Imputation Choice: Median vs. Mean
During the exploratory data analysis (EDA) phase, metrics such as `Insulin` and `SkinThickness` exhibited highly skewed distributions due to clinical outliers (patients with exceptionally high acute values). 

* **The Risk of Mean Imputation:** Utilizing the mean would have skewed the baseline profile upward, artificially inflation-biasing the metrics for typical patients.
* **The Median Solution:** The median was deliberately selected to preserve the central tendency of the distribution without outlier distortion. 

While this defensive imputation strategy successfully protected data variance, it naturally created uniform feature sub-profiles among previously missing records. This explains the 708 value-level profile twins handled in our prior de-duplication step.

```python
#  Scale features (CRITICAL for Logistic Regression)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

```

```python
#Training from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train)

y_pred = log_model.predict(X_test)
accuracy_score(y_test, y_pred)
```

Logistic Regression model correctly predicted diabetes outcomes about:78.61% of the time.

```python
confusion_matrix(y_test, y_pred)
```

The confusion matrix showed that the model correctly identified 434 healthy individuals (True Negatives) and 132 diabetic patients (True Positives). However, 45 healthy individuals were incorrectly classified as diabetic (False Positives), while 109 diabetic patients were incorrectly predicted as healthy (False Negatives). These results indicate that although the model performs well in identifying non-diabetic individuals, it still misses a substantial number of diabetic cases. While predictions of diabetes are reasonably reliable, the relatively high number of false negatives limits the model’s effectiveness for early diabetes screening and risk detection.

```python
print(classification_report(y_test, y_pred))
```

The model performed well in identifying healthy individuals (Class 0 recall = 0.91) and demonstrated moderate performance in detecting diabetic patients (Class 1 recall = 0.55). This indicates that a considerable proportion of at-risk individuals were still incorrectly classified as healthy, which may reduce the model’s effectiveness for diabetes screening. Additionally, the precision (0.75) and F1-score (0.63) for Class 1 suggest moderate predictive performance for diabetic cases, with predictions being reasonably reliable but still missing a notable number of true diabetes cases.

```python
# Evaluate using your 0.30 probability threshold
y_prob = log_reg.predict_proba(X_test_scaled)[:, 1]
y_pred = (y_prob >= 0.30).astype(int)

print("--- Logistic Regression Classification Report (Threshold: 0.30) ---")
print(classification_report(y_test, y_pred))
```

lowering the prediction threshold from the default 0.50 to 0.30 is to make the model more sensitive to diabetic cases, thereby improving recall for Class 1.
