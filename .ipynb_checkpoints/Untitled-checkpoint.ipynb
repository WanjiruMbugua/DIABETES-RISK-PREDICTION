```python
#highly skewed replacing 0 values with median
import numpy as np

columns = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for col in columns:
    Diabetes[col] = Diabetes[col].replace(0, np.nan)
    Diabetes[col] = Diabetes[col].fillna(Diabetes[col].median())
```

TABLE MERGING FOR ANALYSIS




```python
#Rebin age group

def recode_age(age):
    if 18 <= age <= 29:
        return "18-29"
    elif 30 <= age <= 39:
        return "30-39"
    elif 40 <= age <= 49:
        return "40-49"
    elif 50 <= age <= 59:
        return "50-59"
    elif 60 <= age <= 69:
        return "60-69"
    else:
        return "70+"

Diabetes["Age_group"] = Diabetes["Age"].apply(recode_age)
```

FEATURE ENGINEERING DIET SCORE AND NUTRITION SCORE

```python
#Converts diet columns to numeric
diet_cols = [
    "High dietary salt",
    "High dietary sugar",
    "Low fruit intake",
    "Low vegetable intake"
]

clean_Final[diet_cols] = clean_Final[diet_cols].apply(pd.to_numeric, errors="coerce")
```

```python
#cretes diet score 1 to 5
clean_Final["Diet_score"] = (
    (5 - clean_Final["High dietary salt"]) +
    (5 - clean_Final["High dietary sugar"]) +
    (5 - clean_Final["Low fruit intake"]) +
    (5 - clean_Final["Low vegetable intake"])
) / 4
clean_Final["Diet_score"] = clean_Final["Diet_score"].round(2).clip(1,5)
```

```python
#nutrition score

nutri_cols = ["BMI", "Glucose", "BloodPressure", "Insulin"]

clean_Final[nutri_cols] = clean_Final[nutri_cols].apply(pd.to_numeric, errors="coerce")
```

```python
#transforming variables to the same scale 1 = lowest observed risk/value 5 = highest observed risk/value
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(1,5))

scaled = scaler.fit_transform(clean_Final[nutri_cols])

scaled_clean_Final = pd.DataFrame(
    scaled,
    columns=nutri_cols,
    index=clean_Final.index
)
```

```python
clean_Final["Nutrition_score"] = scaled_clean_Final.mean(axis=1).round(2)
```

ANALYTICS MODELLING To predict diabetes

```python
from scipy.stats import ttest_ind

# Variables to test
variables = [
    "Glucose",
    "BMI",
    "Nutrition_score",
    "Diet_score",
    "Age"
]

# Separate groups
diabetic = clean_Final[clean_Final["Outcome"] == 1]
non_diabetic = clean_Final[clean_Final["Outcome"] == 0]

# Run t-tests
for var in variables:

    t_stat, p_value = ttest_ind(
        diabetic[var],
        non_diabetic[var],
        nan_policy='omit'
    )

    print(f"\nVariable: {var}")
    print(f"T-statistic: {t_stat:.3f}")
    print(f"P-value: {p_value:.5f}")
```

Independent sample t-tests were conducted to compare mean values of Glucose, BMI, Nutrition Score, Diet Score, and Age between diabetic and non-diabetic participants. Results showed statistically significant differences for all variables (p < 0.001). Glucose (t = 15.674) and Nutrition Score (t = 15.365) demonstrated the strongest group differences, suggesting a strong association with diabetes outcome.

```python

from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Best threshold (Youden’s J statistic)
j_scores = tpr - fpr
best_threshold = thresholds[np.argmax(j_scores)]

print("Best threshold:", best_threshold)
```
